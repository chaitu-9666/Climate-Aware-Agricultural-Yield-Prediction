# Exploratory analysis — crop yield drivers

Use this notebook to profile real datasets before promoting them to the training pipeline. The cells below assume execution from the **repository root** with `PYTHONPATH` set to `.`.

```python
from pathlib import Path

import pandas as pd
import plotly.express as px

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
csv_path = ROOT / "data" / "raw" / "sample_crop_yields.csv"
df = pd.read_csv(csv_path)
df.head()
```

```python
summary = df.describe().T
summary
```

```python
fig = px.scatter(
    df,
    x="rainfall_mm",
    y="yield_kg_ha",
    color="crop_type",
    marginal_y="histogram",
    title="Yield vs rainfall by crop",
)
fig.show()
```
